# 1. Highest Revenue by Category

```python
category_rev = df.groupby('category')['purchase_amount'].sum().sort_values(ascending=False)

plt.figure(figsize=(10,6))
sns.barplot(x=category_rev.index, y=category_rev.values)

plt.title("Highest Revenue by Category")
plt.xlabel("Category")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.show()
```

# 2. Discounts Increasing Purchase Value

```python
discount_data = df.groupby('discount_applied')['purchase_amount'].agg(['sum','mean']).reset_index()

discount_data.plot(
    x='discount_applied',
    y=['sum','mean'],
    kind='bar',
    figsize=(8,5)
)

plt.title("Discount Impact on Revenue & Avg Purchase")
plt.ylabel("Amount")
plt.show()
```

# 3. Revenue by Gender

```python
gender_rev = df.groupby('gender')['purchase_amount'].sum()

plt.figure(figsize=(6,5))
plt.pie(gender_rev.values,
        labels=gender_rev.index,
        autopct='%1.1f%%')

plt.title("Revenue Contribution by Gender")
plt.show()
```

# 4. High Spending Discount Customers

```python
avg_purchase = df['purchase_amount'].mean()

high_spenders = df[
    (df['discount_applied']=='Yes') &
    (df['purchase_amount'] > avg_purchase)
]

top10 = high_spenders.nlargest(10,'purchase_amount')

plt.figure(figsize=(12,6))
sns.barplot(
    x='customer_id',
    y='purchase_amount',
    data=top10
)

plt.title("Top High-Spending Discount Customers")
plt.xticks(rotation=45)
plt.show()
```

# 5. Top  & Bottom Rated Products

# Top 5

```python
top_products = df.groupby('item_purchased')['review_rating']\
                 .mean()\
                 .sort_values(ascending=False)\
                 .head(5)

plt.figure(figsize=(8,5))
sns.barplot(x=top_products.values, y=top_products.index)

plt.title("Top 5 Highest Rated Products")
plt.xlabel("Average Rating")
plt.show()
```

# Bottom 5

```python
bottom_products = df.groupby('item_purchased')['review_rating']\
                    .mean()\
                    .sort_values()\
                    .head(5)

plt.figure(figsize=(8,5))
sns.barplot(x=bottom_products.values, y=bottom_products.index)

plt.title("Bottom 5 Lowest Rated Products")
plt.xlabel("Average Rating")
plt.show()
```

# 6. Shipping Type vs Revenue

```python
shipping = df.groupby('shipping_type')['purchase_amount'].agg(['mean','sum'])

shipping.plot(kind='bar', figsize=(10,6))

plt.title("Shipping Type vs Revenue")
plt.ylabel("Amount")
plt.show()
```

# 7. Subscribers vs Non-Subscribers

```python
subs = df.groupby('subscription_status')['purchase_amount'].agg(['mean','sum'])

subs.plot(kind='bar', figsize=(8,5))

plt.title("Subscribers vs Non-Subscribers")
plt.ylabel("Revenue")
plt.show()
```

# 8. Top Products with Highest Discount Usage

```python
discount_usage = df.groupby('item_purchased').apply(
    lambda x: (x['discount_applied']=='Yes').mean()*100
).sort_values(ascending=False).head(5)

plt.figure(figsize=(10,5))
sns.barplot(x=discount_usage.values, y=discount_usage.index)

plt.title("Top 5 Products with Highest Discount Usage %")
plt.xlabel("Discount Usage %")
plt.show()
```

# 9. Customer Segmentation

```python
def segment(x):
    if x == 0:
        return 'New Customer'
    elif x <= 15:
        return 'Returning Customer'
    else:
        return 'Loyal Customer'

df['customer_segment'] = df['previous_purchases'].apply(segment)

segment_count = df['customer_segment'].value_counts()

plt.figure(figsize=(7,5))
sns.barplot(x=segment_count.index, y=segment_count.values)

plt.title("Customer Segmentation")
plt.ylabel("Customer Count")
plt.show()
```

# 10. Top Products in Each Category

```python
top_products = (
    df.groupby(['category','item_purchased'])
    .size()
    .reset_index(name='count')
)

top_products = top_products.sort_values(
    ['category','count'],
    ascending=[True,False]
)

top3 = top_products.groupby('category').head(3)

plt.figure(figsize=(14,6))
sns.barplot(
    data=top3,
    x='category',
    y='count',
    hue='item_purchased'
)

plt.title("Top 3 Products in Each Category")
plt.xticks(rotation=45)
plt.show()
```

# 11. Repeat Buyers vs Subscription

```python
df['customer_type'] = df['previous_purchases'].apply(
    lambda x: 'Repeat Buyer' if x > 5 else 'Normal Buyer'
)

repeat_subs = pd.crosstab(
    df['customer_type'],
    df['subscription_status']
)

repeat_subs.plot(kind='bar', stacked=True, figsize=(8,5))

plt.title("Repeat Buyers vs Subscription")
plt.ylabel("Count")
plt.show()
```

# 12. Revenue by Age Group

```python
bins = [18,25,35,50,100]
labels = ['18-25','26-35','36-50','51+']

df['age_group'] = pd.cut(
    df['age'],
    bins=bins,
    labels=labels
)

age_rev = df.groupby('age_group')['purchase_amount'].sum()

plt.figure(figsize=(8,5))
sns.barplot(x=age_rev.index, y=age_rev.values)

plt.title("Revenue Contribution by Age Group")
plt.ylabel("Revenue")
plt.show()
```
